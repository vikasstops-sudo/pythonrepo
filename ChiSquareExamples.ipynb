```python
import numpy as np
from scipy.stats import chi2_contingency

data = np.array([[30 ,40] , [20, 30]])

chi2, p, dof, expected = chi2_contingency(data)

if p<0.05:
    print("Reject")
else:
    print("Fail to reject")
```

## Example

```python
The Chi-Square (χ²) Test is a statistical test used to determine 
whether there is a significant relationship between categorical variables.

It compares:
Observed Values vs Expected Values

Observed Frequency or Value (Given Values)
Expected Values (he count we would expect in a cell if the Null Hypothesis (H₀) were true.)
    
When is Chi-Square Used?
    Test of Independence
        Gender vs Product Preference
        Education vs Job Type
    Goodness of Fit Test
        Does a dice follow a uniform distribution?
        Does customer traffic follow expected proportions?
```
